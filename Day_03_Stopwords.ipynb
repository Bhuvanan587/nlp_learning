```python
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
```

```python
en_stopwords = stopwords.words('english')
```

```python
type(en_stopwords)
```

```python
print(en_stopwords)
```

```python
sentence = "Although technology continues to evolve rapidly, developing a strong foundation in programming and problem-solving will help you adapt to new challenges. Consistency and curiosity are more valuable than trying to learn everything at once."
```

```python
new_sentence = " ".join([word for word in sentence.split() if word not in en_stopwords])
```

```python
print(new_sentence)
```
