```python
import nltk
from nltk.stem import PorterStemmer
```

```python
ps = PorterStemmer()
```

```python
connect_tokens = ["Connection","Connected","Connector","Connect","Connecting"]
```

```python
for t in connect_tokens:
    print(t,": ",ps.stem(t))
```

```python
print(ps.stem("liking"))
```

```python
print(ps.stem("milking"))
```

```python
nltk.download('wordnet')
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
for t in connect_tokens:
    print(t, ": ",lemmatizer.lemmatize(t))
```
